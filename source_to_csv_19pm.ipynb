```python
# imports
import pandas as pd
```

```python
# Ruta relativa del xslx que contiene los datos
original_data_path = 'Data\\Original\\PERSONAS_DEMOGRAFICO_Cuadros_CNPV_2018.xlsx'

# Nombre del csv objetivo
csv_total_filename = 'caldas_viterbo_total_19PM.csv'
csv_filename = 'caldas_viterbo_19PM.csv'

# Ruta relativa del csv objetivo
csv_total_path = 'Data\\Datasets\\' + csv_total_filename
csv_path = 'Data\\Datasets\\' + csv_filename
```

```python
# Lee archivo xlsx
df_19pm = pd.read_excel(original_data_path, sheet_name='19PM', skiprows=10)
df_19pm.head()
```

Se filtrarán los datos que no corresponden al municipio de Viterbo Caldas, para ello se propagará el municipio, el cual se encuentra en la columna "Unnamed: 1" y se seleccionarán solo los registros que contengan el identificador de Viterbo dentro del dataframe, para este caso es "17877_Viterbo".

```python
df_19pm['Unnamed: 1'] = df_19pm['Unnamed: 1'].ffill()
df_19pm = df_19pm.loc[df_19pm['Unnamed: 1'] == '17877_Viterbo']
df_19pm
```

Se eliminan las columnas con datos irrelevantes.

```python
df_19pm.drop(columns=['Unnamed: 0', 'Unnamed: 1'], inplace=True)
df_19pm
```

Se renombran los headers sin nombre.

```python
df_19pm.rename(columns={'Unnamed: 2': 'area', 'Unnamed: 3': 'sexo', 'Unnamed: 4': 'pertenencia_etnica'}, inplace=True)
df_19pm
```

Se propagan los valores de las columnas con valores faltantes.

```python
df_19pm['area'] = df_19pm['area'].ffill()
df_19pm['sexo'] = df_19pm['sexo'].ffill()
df_19pm
```

Se modifican los headers con tildes y ñ.

```python
df_19pm.columns.values
```

```python
headers_replacement = str.maketrans({'-': ' ', 'ó': 'o'})
df_19pm.columns = ["_".join(header.translate(headers_replacement).split()).lower() for header in df_19pm.columns]
df_19pm
```

Se modifican los datos con tildes y ñ.

```python
replacements = str.maketrans({'é': 'e', 'í': 'i', 'ó': 'o', 'ú': 'u'})
df_19pm.loc[:, 'pertenencia_etnica'] = df_19pm['pertenencia_etnica'].str.translate(replacements)
df_19pm
```

Se creará una copia del dataframe y se removerán las filas y columnas con resumenes de datos para el csv que no contiene resúmenes.

```python
df_19pm_cleaned = df_19pm.copy()
df_19pm_cleaned.drop(['total'], axis=1, inplace=True)
df_19pm_cleaned
```

```python
df_19pm_cleaned = df_19pm_cleaned.loc[(df_19pm_cleaned['area'] != 'Total') & (df_19pm_cleaned['sexo'] != 'Total') & (df_19pm_cleaned['pertenencia_etnica'] != 'Total')]
df_19pm_cleaned
```

Los dataframes están listos para ser exportados a CSV.

```python
df_19pm.to_csv(csv_total_path, index=False)
df_19pm_cleaned.to_csv(csv_path, index=False)
```
