```python
# imports
import pandas as pd
```

```python
# Ruta relativa del xslx que contiene los datos
original_data_path = 'Data\\Original\\PERSONAS_DEMOGRAFICO_Cuadros_CNPV_2018.xlsx'

# Nombre del csv objetivo
csv_total_filename = 'caldas_viterbo_total_5_2PM.csv'
csv_filename = 'caldas_viterbo_5_2PM.csv'

# Ruta relativa del csv objetivo
csv_total_path = 'Data\\Datasets\\' + csv_total_filename
csv_path = 'Data\\Datasets\\' + csv_filename
```

```python
# Lee archivo xlsx
df_5_2pm = pd.read_excel(original_data_path, sheet_name='5.2PM', skiprows=9)
df_5_2pm.head()
```

Se observa que los headers se encuentran en la primera fila, para ellos se creará un diccionario con el nombre actual de la columna como llave y el nombre correspondiente de la primera fila como valor, para luego reemplazarlos.

```python
df_new_headers = {}
current_headers = list(df_5_2pm.columns.values)[6:]
actual_headers = list(df_5_2pm.iloc[0])[6:]

for key, value in zip(current_headers, actual_headers):
    df_new_headers[key] = value

df_5_2pm = df_5_2pm.rename(columns=df_new_headers)
df_5_2pm
```

Se cambian los nombres de las columnas restantes por nombres más descriptivos.

```python
df_5_2pm = df_5_2pm.rename(columns={'Unnamed: 1':'Municipio', 'Unnamed: 2':'Sexo', 'Unnamed: 3':'Area', 'Unnamed: 4':'Grupo de edad'})
df_5_2pm
```

Se propagan los identificadores de municipio para filtrar los datos que no corresponden a Viterbo, Caldas.

```python
df_5_2pm['Municipio'] = df_5_2pm['Municipio'].ffill()
df_5_2pm = df_5_2pm.loc[df_5_2pm['Municipio'] == '17877_Viterbo']
df_5_2pm
```

Se eliminan columnas que no contienen datos relevantes para el dataset y aquellas que contienen resumenes de datos.

```python
list(df_5_2pm.columns.values)
```

```python
df_5_2pm = df_5_2pm.drop(['sexo, grupos de edad y áreas (Total, Cabecera y Centros poblados y Rural disperso), sexo y grupos de edad',
               'Municipio'], axis=1)
df_5_2pm
```

Se renombran los headers que tienen tildes y ñ.

```python
df_5_2pm = df_5_2pm.rename(columns={'Pareja (Conyuge, compañero(a), esposo(a))':'Pareja (Conyuge, companero(a), esposo(a))',
                         'Cuñado(a)':'Cunado(a)',
                         'Empleado(a) del servicio doméstico':'Empleado(a) del servicio domestico'})
df_5_2pm
```

Se propagan los valores del area y sexo para reemplazar valores NaN.

```python
df_5_2pm[['Sexo', 'Area']] = df_5_2pm[['Sexo', 'Area']].ffill()
df_5_2pm
```

Se reemplazan las tildes dentro de los datos.

```python
df_5_2pm['Grupo de edad'] = df_5_2pm['Grupo de edad'].str.replace('á', 'a')
df_5_2pm
```

Se creará una copia del dataframe y se removerán las filas y columnas con resumenes de datos para el csv que no contiene resúmenes.

```python
df_5_2pm_cleaned = df_5_2pm.copy()
df_5_2pm_cleaned.drop(['Total'], axis=1)
df_5_2pm_cleaned
```

```python
df_5_2pm_cleaned = df_5_2pm_cleaned.loc[(df_5_2pm_cleaned['Area'] != 'Total') & (df_5_2pm_cleaned['Grupo de edad'] != 'Ttoal')]
df_5_2pm_cleaned
```

Los dataframes están listos para ser exportados a CSV.

```python
df_5_2pm.to_csv(csv_total_path, index=False)
df_5_2pm_cleaned.to_csv(csv_path, index=False)
```
