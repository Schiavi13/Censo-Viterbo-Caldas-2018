```python
# imports
import pandas as pd
```

```python
# Ruta relativa del xslx que contiene los datos
original_data_path = 'Data\\Original\\PERSONAS_DEMOGRAFICO_Cuadros_CNPV_2018.xlsx'

# Nombre del csv objetivo
csv_filename = 'caldas_viterbo_1PM.csv'

# Ruta relativa del csv objetivo
csv_path = 'Data\\Datasets\\' + csv_filename
```

```python
# Lee archivo xlsx
df_1PM = pd.read_excel(original_data_path, sheet_name='1PM', skiprows=9)
df_1PM.head()
```

```python
list(df_1PM.columns.values)
```

Se reajustan los headers para que sean más descriptivos y para que 'Hombre' y 'Mujer' sean parte de estos.
A simple vista se observa que los headers 'Unnamed: 2' y 'Unnamed: 3' corresponden a Áreas y Grupos de edad respecrivamente. También se eliminarán las tildes de los headers existentes.

```python
df_1PM = df_1PM.rename(columns={'Unnamed: 2': 'Area','Unnamed: 3': 'Grupo de edad','Sexo':'Hombre', 'Unnamed: 6':'Mujer', 
                                'Índice de  masculinidad':'Indice de  masculinidad', 'Índice de  feminidad':'Indice de  feminidad'})
list(df_1PM.columns.values)
```

```python
df_1PM.iloc[24185:24242]
```

Se observa que el header 'Unnamed: 1' corresponde al municipio, y la primera columna no contiene información importante para el dataset, por lo tanto se pueden eliminar.

```python
df_1PM = df_1PM.iloc[24185:24242]
df_1PM = df_1PM.drop(['Grupos de edad, municipio y áreas  (Total, Cabecera y Centros poblados y Rural disperso) ', 'Unnamed: 1'], axis=1)
```

```python
df_1PM
```

Los valores NaN en la columna 'Area' deben ser reemplazados por el primer valor válido de la columna hasta el siguiente valor válido, es decir, propagar el valor hasta encontrar el siguiente.

```python
df_1PM['Area'] = df_1PM['Area'].ffill()
df_1PM
```

Se reemplazarán las tildes del contenido, en este caso específico, solo la columna 'Grupo de edad contiene datos con tildes.

```python
df_1PM['Grupo de edad'] = df_1PM['Grupo de edad'].str.replace('á', 'a')
df_1PM
```

El DataFrame está listo para ser exportado a csv

```python
df_1PM.to_csv(csv_path, index=False)
```
