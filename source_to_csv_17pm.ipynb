```python
# imports
import pandas as pd
```

```python
# Ruta relativa del xslx que contiene los datos
original_data_path = 'Data\\Original\\PERSONAS_DEMOGRAFICO_Cuadros_CNPV_2018.xlsx'

# Nombre del csv objetivo
csv_total_filename = 'caldas_viterbo_total_17PM.csv'
csv_filename = 'caldas_viterbo_17PM.csv'

# Ruta relativa del csv objetivo
csv_total_path = 'Data\\Datasets\\' + csv_total_filename
csv_path = 'Data\\Datasets\\' + csv_filename
```

```python
# Lee archivo xlsx
df_17pm = pd.read_excel(original_data_path, sheet_name='17PM', skiprows=10)
df_17pm.head()
```

Se filtrarán los datos que no corresponden al municipio de Viterbo Caldas, para ello se propagará el municipio, el cual se encuentra en la columna "Unnamed: 1" y se seleccionarán solo los registros que contengan el identificador de Viterbo dentro del dataframe, para este caso es "17877_Viterbo".

```python
df_17pm['Unnamed: 1'] = df_17pm['Unnamed: 1'].ffill()
df_17pm = df_17pm.loc[df_17pm['Unnamed: 1'] == '17877_Viterbo']
df_17pm
```

Se eliminan las columnas con datos irrelevantes.

```python
df_17pm.drop(columns=['Unnamed: 0', 'Unnamed: 1'], inplace=True)
df_17pm
```

Se renombran los headers sin nombre.

```python
df_17pm.rename(columns={'Unnamed: 2': 'Area', 'Unnamed: 3': 'Sexo'}, inplace=True)
df_17pm
```

Se propagan los valores de las columnas con valores faltantes.

```python
df_17pm['Area'] = df_17pm['Area'].ffill()
df_17pm
```

Se modifican los headers que tienen tildes.

```python
df_17pm.columns.values
```

```python
df_17pm.rename(columns={'Especialización, Maestria, Doctorado': 'Especializacion, Maestria, Doctorado', 'Sin Información ': 'Sin informacion' }, inplace=True)
df_17pm
```

Se creará una copia del dataframe y se removerán las filas y columnas con resumenes de datos para el csv que no contiene resúmenes.

```python
df_17pm_cleaned = df_17pm.copy()
df_17pm_cleaned.drop(['Total'], axis=1, inplace=True)
df_17pm_cleaned
```

```python
df_17pm_cleaned = df_17pm_cleaned.loc[(df_17pm_cleaned['Area'] != 'Total') & (df_17pm_cleaned['Sexo'] != 'Total')]
df_17pm_cleaned
```

Los dataframes están listos para ser exportados a CSV.

```python
df_17pm.to_csv(csv_total_path, index=False)
df_17pm_cleaned.to_csv(csv_path, index=False)
```
