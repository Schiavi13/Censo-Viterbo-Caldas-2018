```python
# imports
import pandas as pd
```

```python
# Ruta relativa del xslx que contiene los datos
original_data_path = 'Data\\Original\\PERSONAS_DEMOGRAFICO_Cuadros_CNPV_2018.xlsx'

# Nombre del csv objetivo
csv_filename = 'caldas_viterbo_14PM.csv'

# Ruta relativa del csv objetivo
csv_path = 'Data\\Datasets\\' + csv_filename
```

```python
# Lee archivo xlsx
df_14pm = pd.read_excel(original_data_path, sheet_name='14PM', skiprows=8)
df_14pm.head()
```

```python
first_row = list(df_14pm.iloc[0])[:]
for (i, value) in enumerate(first_row):
    print(i, value)
```

```python
second_row = list(df_14pm.iloc[1])[:]
for (i, value) in enumerate(second_row):
    print(i, value)
```

```python
third_row = list(df_14pm.iloc[2])[:]
for (i, value) in enumerate(third_row):
    print(i, value)
```

Se observan secciones altamente agrupadas en columnas con subcolumnas, y estas subcolumnas tienen a su vez otras subcolumnas.
El primer subgrupo de columnas correspondientes a "Tuvo alguna enfermedad en los últimos 30 días" y "Servicio de salud donde acudieron" se puede eliminar al modificar algunos nombres de las subcolumnas con mayor granularidad y documentar. En este caso, si se utilizan las columnas con más granularidad, se tienen dos columnas con el mismo nombre "Sin información" y que no se pueden ignorar en el proceso de extracción de datos.

```python
actual_headers = list(df_14pm.iloc[2][5:19])
del actual_headers[3]
actual_headers[2] = "Sin informacion enfermedad ultimos 30 dias"
actual_headers[4] = "Sin informacion servicio de salud donde acudieron"
actual_headers
```

Se extraen las columnas con mayor granularidad.

Se filtrarán los datos que no corresponden al municipio de Viterbo Caldas, para ello se propagará el municipio, el cual se encuentra en la columna "Unnamed: 1" y se seleccionarán solo los registros que contengan el identificador de Viterbo dentro del dataframe, para este caso es "17877_Viterbo".

```python
df_14pm['Unnamed: 1'] = df_14pm['Unnamed: 1'].ffill()
df_14pm = df_14pm.loc[df_14pm['Unnamed: 1'] == '17877_Viterbo']
df_14pm
```

Debido a la agrupación de columnas por área, se separan los datos por subconjuntos dependiendo del área y al final serán reunidos en un mismo dataframe con el área presente en las filas, se eliminan columnas con resumenes de datos, aka totales.

```python
ls_df = []
area_type = ['cabecera', 'centro_poblado', 'rural_disperso']
for i, area in zip(range(22, 70, 17), area_type):
    df_14pm_with_area = pd.concat([df_14pm[['Unnamed: 2', 'Unnamed: 3']], df_14pm.iloc[:,i:i+14].assign(Area=area)], axis=1)
    df_14pm_with_area.drop(df_14pm_with_area.columns[5], axis=1, inplace=True)
    ls_df.append(df_14pm_with_area)

for df in ls_df:
    df.rename(columns=dict(zip(list(df.columns.values[2:-1]), actual_headers[:])), inplace=True)

df_14pm = pd.concat(ls_df)
df_14pm.rename(columns={'Unnamed: 2': 'Sexo', 'Unnamed: 3': 'Grupo de edad'}, inplace=True)
df_14pm
```

Se propaga el valor de la columna 'Sexo' para completar los datos faltantes.

```python
df_14pm['Sexo'] = df_14pm['Sexo'].ffill()
df_14pm
```

Se renombran los headers que tienen tildes y ñ.

```python
df_14pm.rename(columns={'Sí': 'Si',
 'A la entidad de seguridad social en salud a la cual está afliado(a)': 'A la entidad de seguridad social en salud a la cual esta afliado(a)',
 'A un médico particular': 'A un medico particular',
 'A un boticario,farmacéuta, droguista': 'A un boticario,farmaceuta, droguista',
 'Acudió a una autoridad indígena espiritual': 'Acudio a una autoridad indigena espiritual',
 'Otro médico de un grupo étnico': 'Otro medico de un grupo etnico',
 'Usó remedios caseros': 'Uso remedios caseros',
 'Se autorrecetó': 'Se autorreceto'}, inplace=True)

df_14pm
```

Se cambian los datos que contienen tildes y ñ.

```python
replacements = str.maketrans({'ñ':'ni', 'á':'a'})
df_14pm.loc[:,'Grupo de edad'] = df_14pm['Grupo de edad'].str.translate(replacements)
df_14pm
```

El dataframe está listo para ser exportado a CSV.

```python
df_14pm.to_csv(csv_path, index=False)
```
