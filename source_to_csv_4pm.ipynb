```python
# imports
import pandas as pd
```

```python
# Ruta relativa del xslx que contiene los datos
original_data_path = 'Data\\Original\\PERSONAS_DEMOGRAFICO_Cuadros_CNPV_2018.xlsx'

# Nombre del csv objetivo
csv_filename = 'caldas_viterbo_4PM.csv'

# Ruta relativa del csv objetivo
csv_path = 'Data\\Datasets\\' + csv_filename
```

```python
# Lee archivo xlsx
df_4pm = pd.read_excel(original_data_path, sheet_name='4PM', skiprows=9)
df_4pm.head()
```

Se observa que los headers se encuentran en la primera fila, para ellos se creará un diccionario con el nombre actual de la columna como llave y el nombre correspondiente de la primera fila como valor, para luego reemplazarlos. 


```python
df_new_headers = {}
current_headers = list(df_4pm.columns.values)[5:]
actual_headers = list(df_4pm.iloc[0])[5:]

for key, value in zip(current_headers, actual_headers):
    df_new_headers[key] = value

df_4pm = df_4pm.rename(columns=df_new_headers)
df_4pm
```

Se cambian los nombres de las columnas restantes por nombres más descriptivos.

```python
df_4pm = df_4pm.rename(columns={'Unnamed: 1':'Municipio', 'Unnamed: 2':'Area', 'Unnamed: 3':'Sexo'})
df_4pm
```

Se propagan los identificadores de municipio para filtrar los datos que no corresponden a Viterbo, Caldas.

```python
df_4pm['Municipio'] = df_4pm['Municipio'].ffill()
df_4pm = df_4pm.loc[df_4pm['Municipio'] == '17877_Viterbo']
df_4pm
```

Se eliminan columnas que no contienen datos relevantes para el dataset y aquellas que contienen resumenes de datos.

```python
df_4pm = df_4pm.drop(['Municipio, área (Total, Cabecera y Centros poblados y Rural disperso) y sexo', 'Municipio'], axis=1)
df_4pm
```

Se renombran los headers que tienen tildes y ñ.

```python
df_4pm = df_4pm.rename(columns={'Pareja (Conyuge, compañero(a), esposo(a))':'Pareja (Conyuge, companero(a), esposo(a))',
                       'Cuñado(a)':'Cunado(a)',
                       'Empleado(a) del servicio doméstico':'Empleado(a) del servicio domestico'})
df_4pm
```

Se propaga el area para reemplazar valores NaN.

```python
df_4pm['Area'] = df_4pm['Area'].ffill()
df_4pm
```

El dataframe está listo para ser exportado a csv.

```python
df_4pm.to_csv(csv_path, index=False)
```
