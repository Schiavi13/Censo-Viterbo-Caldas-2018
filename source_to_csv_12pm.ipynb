```python
# imports
import pandas as pd
```

```python
# Ruta relativa del xslx que contiene los datos
original_data_path = 'Data\\Original\\PERSONAS_DEMOGRAFICO_Cuadros_CNPV_2018.xlsx'

# Nombre del csv objetivo
csv_filename = 'caldas_viterbo_12PM.csv'

# Ruta relativa del csv objetivo
csv_path = 'Data\\Datasets\\' + csv_filename
```

```python
# Lee archivo xlsx
df_12pm = pd.read_excel(original_data_path, sheet_name='12PM', skiprows=9)
df_12pm.head()
```

Ya que las columnas están agrupadas por área, y es necesario modificar los headers, se extraerán los headers correctos en una lista y se utilizarán apra modificar los headers actuales en un paso futuro cuando se tengan conjuntos de datos más reducidos.

```python
first_row = list(df_12pm.iloc[0])[:]
for (i, value) in enumerate(first_row):
    print(i, value)
```

```python
actual_headers = list(df_12pm.iloc[0][4:11])
actual_headers
```

Se filtrarán los datos que no corresponden al municipio de Viterbo Caldas, para ello se propagará el municipio, el cual se encuentra en la columna "Unnamed: 1" y se seleccionarán solo los registros que contengan el identificador de Viterbo dentro del dataframe, para este caso es "17877_Viterbo".

```python
df_12pm['Unnamed: 1'] = df_12pm['Unnamed: 1'].ffill()
df_12pm = df_12pm.loc[df_12pm['Unnamed: 1'] == '17877_Viterbo']
df_12pm
```

Debido a la agrupación de columnas por área, se separan los datos por subconjuntos dependiendo del área y al final serán reunidos en un mismo dataframe con el área presente en las filas, se eliminan columnas con resumenes de datos, aka totales.

```python
# Pendiente de automatizar con loops, resolver problemas de manipulación de dataframes dentro de for loops
ls_df = []
df_12pm_grupo_edad = df_12pm['Unnamed: 2']
df_12pm_cabecera = pd.concat([df_12pm_grupo_edad, df_12pm.iloc[:,13:20]], axis=1)
ls_df.append(df_12pm_cabecera)
df_12pm_centros_poblados = pd.concat([df_12pm_grupo_edad, df_12pm.iloc[:,22:29]], axis=1)
ls_df.append(df_12pm_centros_poblados)
df_12pm_rural_disperso = pd.concat([df_12pm_grupo_edad, df_12pm.iloc[:,31:]], axis=1)
ls_df.append(df_12pm_rural_disperso)
ls_df
df_12pm_cabecera = df_12pm_cabecera.rename(columns={'Unnamed: 2':'Grupo edad'})
df_12pm_centros_poblados = df_12pm_centros_poblados.rename(columns={'Unnamed: 2':'Grupo edad'})
df_12pm_rural_disperso = df_12pm_rural_disperso.rename(columns={'Unnamed: 2':'Grupo edad'})
df_12pm_cabecera.columns.values[1:] = actual_headers[:]
df_12pm_centros_poblados.columns.values[1:] = actual_headers[:]
df_12pm_rural_disperso.columns.values[1:] = actual_headers[:]
df_12pm_cabecera = df_12pm_cabecera.assign(Area='cabecera')
df_12pm_centros_poblados = df_12pm_centros_poblados.assign(Area='centro poblado')
df_12pm_rural_disperso = df_12pm_rural_disperso.assign(Area='rural disperso')
df_12pm = pd.concat([df_12pm_cabecera, df_12pm_centros_poblados, df_12pm_rural_disperso])
df_12pm
```

Se renombran los headers que tienen tildes y ñ.

```python
df_12pm = df_12pm.rename(columns={'Indígena':'Indigena', 
                        'Raizal del Archipielago de San Andrés, Providencia y Santa Catalina':'Raizal del Archipielago de San Andres, Providencia y Santa Catalina', 
                        'Ningún grupo étnico':'Ningun grupo etnico',
                        'Sin información':'Sin informacion'})
df_12pm
```

Se eliminan las filas que contienen resumenes de datos.

```python
df_12pm = df_12pm.loc[df_12pm['Grupo edad'] != 'Total']
df_12pm
```

```python
replacements = str.maketrans({'ñ':'ni', 'á':'a'})
df_12pm.loc[:,'Grupo edad'] = df_12pm['Grupo edad'].str.translate(replacements)
df_12pm
```

El dataframe está listo para ser exportado a csv.

```python
df_12pm.to_csv(csv_path, index=False)
```
