```python
# imports
import pandas as pd
```

```python
# Ruta relativa del xslx que contiene los datos
original_data_path = 'Data\\Original\\PERSONAS_DEMOGRAFICO_Cuadros_CNPV_2018.xlsx'

# Nombre del csv objetivo
csv_total_filename = 'caldas_viterbo_total_15PM.csv'
csv_filename = 'caldas_viterbo_15PM.csv'

# Ruta relativa del csv objetivo
csv_total_path = 'Data\\Datasets\\' + csv_total_filename
csv_path = 'Data\\Datasets\\' + csv_filename
```

```python
# Lee archivo xlsx
df_15pm = pd.read_excel(original_data_path, sheet_name='15PM', skiprows=9)
df_15pm.head()
```

```python
first_row = list(df_15pm.iloc[0])[:]
for (i, value) in enumerate(first_row):
    print(i, value)
```

Se extraen los headers correctos, y se renombran columnas que no se logran capturar.

```python
actual_headers = list(df_15pm.iloc[0][2:])
actual_headers[0] = 'Area'
actual_headers[1] = 'Sexo'
actual_headers[2] = 'Edad'
actual_headers[3] = 'Total personas menores de 5 anios'
actual_headers
```

Se filtrarán los datos que no corresponden al municipio de Viterbo Caldas, para ello se propagará el municipio, el cual se encuentra en la columna "Unnamed: 1" y se seleccionarán solo los registros que contengan el identificador de Viterbo dentro del dataframe, para este caso es "17877_Viterbo".

```python
df_15pm['Unnamed: 1'] = df_15pm['Unnamed: 1'].ffill()
df_15pm = df_15pm.loc[df_15pm['Unnamed: 1'] == '17877_Viterbo']
df_15pm
```

Se eliminan las columnas con datos no importantes.

```python
df_15pm = df_15pm.drop(df_15pm.columns[:2], axis=1)
df_15pm
```

Se asignan los headers.

```python
df_15pm.rename(columns=dict(zip(list(df_15pm.columns.values), actual_headers)), inplace=True)
df_15pm
```

Se propagan los datos de las columnas 'Area' y 'Sexo' para completar los datos faltantes.

```python
df_15pm['Area'] = df_15pm['Area'].ffill()
df_15pm['Sexo'] = df_15pm['Sexo'].ffill()
df_15pm
```

Se renombran los headers que tienen tildes y ñ.

```python
df_15pm.rename(columns={
    'Asiste a un hogar comunitario, jardín, centro de desarrollo infantil o colegio': 'Asiste a un hogar comunitario, jardin, centro de desarrollo infantil o colegio',
    'En la vivienda donde vive el niño al cuidado de un pariente o una persona de 18 años o más': 'En la vivienda donde vive el ninioo al cuidado de un pariente o una persona de 18 anios o mas',
    'En la vivienda donde vive el niño al cuidado de un pariente o una persona menor de 18 años': 'En la vivienda donde vive el ninio al cuidado de un pariente o una persona menor de 18 anios',
    'Sin Información': 'Sin Informacion'
}, inplace=True)

df_15pm
```

Se creará una copia del dataframe y se removerán las filas y columnas con resumenes de datos para el csv que no contiene resúmenes.

```python
df_15pm_cleaned = df_15pm.copy()
df_15pm_cleaned.drop(['Total personas menores de 5 anios'], axis=1, inplace=True)
df_15pm_cleaned
```

```python
df_15pm_cleaned = df_15pm_cleaned.loc[(df_15pm_cleaned['Area'] != 'Total Departamental') & (df_15pm_cleaned['Sexo'] != 'Total') & (df_15pm_cleaned['Edad'] != 'Total')]
df_15pm_cleaned
```

Los dataframes están listos para ser exportados a CSV.

```python
df_15pm.to_csv(csv_total_path, index=False)
df_15pm_cleaned.to_csv(csv_path, index=False)
```
