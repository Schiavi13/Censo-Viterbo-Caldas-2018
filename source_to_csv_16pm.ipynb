```python
# imports
import pandas as pd
```

```python
# Ruta relativa del xslx que contiene los datos
original_data_path = 'Data\\Original\\PERSONAS_DEMOGRAFICO_Cuadros_CNPV_2018.xlsx'

# Nombre del csv objetivo
csv_total_filename = 'caldas_viterbo_total_16PM.csv'
csv_filename = 'caldas_viterbo_16PM.csv'

# Ruta relativa del csv objetivo
csv_total_path = 'Data\\Datasets\\' + csv_total_filename
csv_path = 'Data\\Datasets\\' + csv_filename
```

```python
# Lee archivo xlsx
df_16pm = pd.read_excel(original_data_path, sheet_name='16PM', skiprows=9)
df_16pm.head()
```

```python
first_row = list(df_16pm.iloc[0])[:]
for (i, value) in enumerate(first_row):
    print(i, value)
```

```python
second_row = list(df_16pm.iloc[1])[:]
for (i, value) in enumerate(second_row):
    print(i, value)
```

Se observan columnas y subcolumnas, se dividirán en diferentes datasets, se asignará una nueva columna "Sabe leer y escribir" y se asignaran los valores correspondientes desagrupando el dataframe en este item, y se unirán los diferentes dataframes.

```python
actual_headers = list(df_16pm.iloc[1][4:])
cleaned_headers = [x for x in actual_headers if not pd.isnull(x)]
cleaned_headers
```

Se filtrarán los datos que no corresponden al municipio de Viterbo Caldas, para ello se propagará el municipio, el cual se encuentra en la columna "Unnamed: 1" y se seleccionarán solo los registros que contengan el identificador de Viterbo dentro del dataframe, para este caso es "17877_Viterbo".

```python
df_16pm['Unnamed: 1'] = df_16pm['Unnamed: 1'].ffill()
df_16pm = df_16pm.loc[df_16pm['Unnamed: 1'] == '17877_Viterbo']
df_16pm
```

Debido a la agrupación de columnas por capacidad de leer y escribir, se separan los datos por subconjuntos dependiendo si los censados saben leer y escribir y al final serán reunidos en un mismo dataframe con el valor presente en las filas.

```python
ls_df = []
literate = ['total', 'si', 'no', 'sin_informacion']
for i, literate_status in zip(range(4, 19, 4), literate):
    df_16pm_with_literate_status = pd.concat([df_16pm[['Unnamed: 2', 'Unnamed: 3']], df_16pm.iloc[:,i:i+3].assign(Sabe_leer_y_escribir=literate_status)], axis=1)
    ls_df.append(df_16pm_with_literate_status)

for df in ls_df:
    df.rename(columns=dict(zip(list(df.columns.values[2:-1]), cleaned_headers[:])), inplace=True)

df_16pm = pd.concat(ls_df)
df_16pm.rename(columns={'Unnamed: 2': 'Area', 'Unnamed: 3': 'Grupo de edad'}, inplace=True)

df_16pm
```

Se propaga el valor de la columna 'Area' para completar los datos faltantes.

```python
df_16pm['Area'] = df_16pm['Area'].ffill()
df_16pm
```

Se cambian los datos que contienen tildes y ñ.

```python
replacements = str.maketrans({'ñ':'ni', 'á':'a'})
df_16pm.loc[:,'Grupo de edad'] = df_16pm['Grupo de edad'].str.translate(replacements)
df_16pm
```

Se creará una copia del dataframe y se removerán las filas y columnas con resumenes de datos para el csv que no contiene resúmenes.

```python
df_16pm_cleaned = df_16pm.copy()
df_16pm_cleaned.drop(['Total'], axis=1, inplace=True)
df_16pm_cleaned
```

```python
df_16pm_cleaned = df_16pm_cleaned.loc[(df_16pm_cleaned['Area'] != 'Total') & (df_16pm_cleaned['Sabe_leer_y_escribir'] != 'total')]
df_16pm_cleaned
```

Los dataframes están listos para ser exportados a CSV.

```python
df_16pm.to_csv(csv_total_path, index=False)
df_16pm_cleaned.to_csv(csv_path, index=False)
```
