```python
# imports
import pandas as pd
```

```python
# Ruta relativa del xslx que contiene los datos
original_data_path = 'Data\\Original\\PERSONAS_DEMOGRAFICO_Cuadros_CNPV_2018.xlsx'

# Nombre del csv objetivo
csv_filename = 'caldas_viterbo_13PM.csv'

# Ruta relativa del csv objetivo
csv_path = 'Data\\Datasets\\' + csv_filename
```

```python
# Lee archivo xlsx
df_13pm = pd.read_excel(original_data_path, sheet_name='13PM', skiprows=9)
df_13pm.head()
```

Ya que las columnas están agrupadas por área, y es necesario modificar los headers, se extraerán los headers correctos en una lista y se utilizarán apra modificar los headers actuales en un paso futuro cuando se tengan conjuntos de datos más reducidos.

```python
first_row = list(df_13pm.iloc[0])[:]
for (i, value) in enumerate(first_row):
    print(i, value)
```

```python
actual_headers = list(df_13pm.iloc[0][4:11])
actual_headers
```

Se filtrarán los datos que no corresponden al municipio de Viterbo Caldas, para ello se propagará el municipio, el cual se encuentra en la columna "Unnamed: 1" y se seleccionarán solo los registros que contengan el identificador de Viterbo dentro del dataframe, para este caso es "17877_Viterbo".

```python
df_13pm['Unnamed: 1'] = df_13pm['Unnamed: 1'].ffill()
df_13pm = df_13pm.loc[df_13pm['Unnamed: 1'] == '17877_Viterbo']
df_13pm
```

Debido a la agrupación de columnas por área, se separan los datos por subconjuntos dependiendo del área y al final serán reunidos en un mismo dataframe con el área presente en las filas, se eliminan columnas con resumenes de datos, aka totales.

```python
ls_df = []
df_13pm_territorio_etnico = df_13pm['Unnamed: 2']
df_13pm_cabecera = pd.concat([df_13pm_territorio_etnico, df_13pm.iloc[:,13:20].assign(Area='cabecera')], axis=1)
ls_df.append(df_13pm_cabecera)
df_13pm_centros_poblados = pd.concat([df_13pm_territorio_etnico, df_13pm.iloc[:,22:29].assign(Area='centro_poblado')], axis=1)
ls_df.append(df_13pm_centros_poblados)
df_13pm_rural_disperso = pd.concat([df_13pm_territorio_etnico, df_13pm.iloc[:,31:].assign(Area='rural_disperso')], axis=1)
ls_df.append(df_13pm_rural_disperso)
for df in ls_df:
    df.rename(columns={'Unnamed: 2':'Territorio etnico'}, inplace=True)
    df.rename(columns=dict(zip(list(df.columns.values[1:-1]), actual_headers[:])), inplace=True)
df_13pm = pd.concat(ls_df)
df_13pm
```

Se renombran los headers que tienen tildes y ñ.

```python
df_13pm = df_13pm.rename(columns={'Indígena':'Indigena', 
                        'Raizal del Archipielago de San Andrés, Providencia y Santa Catalina':'Raizal del Archipielago de San Andres, Providencia y Santa Catalina', 
                        'Ningún grupo étnico':'Ningun grupo etnico',
                        'Sin información':'Sin informacion'})
df_13pm
```

Se eliminan las filas que contienen resumenes de datos.

```python
df_13pm = df_13pm.loc[df_13pm['Territorio etnico'] != 'Total']
df_13pm
```

```python
replacements = str.maketrans({'é':'e', 'í':'i'})
df_13pm.loc[:,'Territorio etnico'] = df_13pm['Territorio etnico'].str.translate(replacements)
df_13pm
```

El dataframe está listo para ser exportado a csv.

```python
df_13pm.to_csv(csv_path, index=False)
```
