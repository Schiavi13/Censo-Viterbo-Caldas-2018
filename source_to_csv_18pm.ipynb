```python
# imports
import pandas as pd
```

```python
# Ruta relativa del xslx que contiene los datos
original_data_path = 'Data\\Original\\PERSONAS_DEMOGRAFICO_Cuadros_CNPV_2018.xlsx'

# Nombre del csv objetivo
csv_total_filename = 'caldas_viterbo_total_18PM.csv'
csv_filename = 'caldas_viterbo_18PM.csv'

# Ruta relativa del csv objetivo
csv_total_path = 'Data\\Datasets\\' + csv_total_filename
csv_path = 'Data\\Datasets\\' + csv_filename
```

```python
# Lee archivo xlsx
df_18pm = pd.read_excel(original_data_path, sheet_name='18PM', skiprows=9)
df_18pm.head()
```

```python
first_row = list(df_18pm.iloc[0])[:]
for (i, value) in enumerate(first_row):
    print(i, value)
```

```python
second_row = list(df_18pm.iloc[1])[:]
for (i, value) in enumerate(second_row):
    print(i, value)
```

Se observan columnas y subcolumnas, se dividirán en diferentes datasets, se asignará una nueva columna "Asistencia a alguna institucion educativa" y se asignaran los valores correspondientes desagrupando el dataframe en este item, y se unirán los diferentes dataframes.

```python
actual_headers = list(df_18pm.iloc[1][4:])
cleaned_headers = [x.lower() for x in actual_headers if not pd.isnull(x)]
cleaned_headers
```

Se filtrarán los datos que no corresponden al municipio de Viterbo Caldas, para ello se propagará el municipio, el cual se encuentra en la columna "Unnamed: 1" y se seleccionarán solo los registros que contengan el identificador de Viterbo dentro del dataframe, para este caso es "17877_Viterbo".

```python
df_18pm['Unnamed: 1'] = df_18pm['Unnamed: 1'].ffill()
df_18pm = df_18pm.loc[df_18pm['Unnamed: 1'] == '17877_Viterbo']
df_18pm
```

Debido a la agrupación de columnas por asistencia a alguna institución educativa, se separan los datos por subconjuntos dependiendo si los censados asisten a alguna institución educativa y al final serán reunidos en un mismo dataframe con el valor presente en las filas.

```python
ls_df = []
academic_institution_atendee = ['Total', 'Si', 'No', 'Sin informacion']
for i, atendee in zip(range(4, 19, 4), academic_institution_atendee):
    df_18pm_with_academic_atendance = pd.concat([df_18pm[['Unnamed: 2', 'Unnamed: 3']], df_18pm.iloc[:, i:i+3].assign(asiste_a_alguna_institucion_educativa=atendee)], axis=1)
    ls_df.append(df_18pm_with_academic_atendance)

for df in ls_df:
    df.rename(columns=dict(zip(list(df.columns.values[2:-1]), cleaned_headers[:])), inplace=True)

df_18pm = pd.concat(ls_df)
df_18pm.rename(columns={'Unnamed: 2': 'area', 'Unnamed: 3': 'grupo_de_edad'}, inplace=True)
df_18pm
```

Se propaga el valor de la columna 'Area' para completar los datos faltantes.

```python
df_18pm['area'] = df_18pm['area'].ffill()
df_18pm
```

Se reemplazan los datos que contienen tildes y ñ.

```python
replacements = str.maketrans({'ñ': 'ni', 'á': 'a'})
df_18pm.loc[:, 'grupo_de_edad'] = df_18pm['grupo_de_edad'].str.translate(replacements)
df_18pm
```

Se creará una copia del dataframe y se removerán las filas y columnas con resumenes de datos para el csv que no contiene resúmenes.

```python
df_18pm_cleaned = df_18pm.copy()
df_18pm_cleaned.drop(['total'], axis=1, inplace=True)
df_18pm_cleaned
```

```python
df_18pm_cleaned = df_18pm_cleaned.loc[(df_18pm_cleaned['area'] != 'Total') & (df_18pm_cleaned['asiste_a_alguna_institucion_educativa'] != 'Total')]
df_18pm_cleaned
```

Los dataframes están listos para ser exportados a CSV.

```python
df_18pm.to_csv(csv_total_path, index=False)
df_18pm_cleaned.to_csv(csv_path, index=False)
```
